Feature Engineering & Selection


```python
df['Order Date']=pd.to_datetime(df['Order Date'])

```

```python
df['Year'] = df['Order Date'].dt.year
```

```python
df['Month'] = df['Order Date'].dt.month
```

```python
df['Day'] = df['Order Date'].dt.day
```

```python
df['Quater'] = df['Order Date'].dt.quarter
```

```python
df['Day Of Week'] = df['Order Date'].dt.day_name()
```

DATA ENCODING


```python
from sklearn.preprocessing import LabelEncoder
```

```python
df.select_dtypes(include='object').columns
```

```python
label_encoder=LabelEncoder()
bin_cols=['Ship Date', 'Ship Mode', 'Customer ID',
       'Customer Name', 'Segment', 'Country', 'City', 'State', 'Region',
       'Product ID', 'Category', 'Sub-Category', 'Product Name']

```

```python
for col in bin_cols:
  df[col] = df[col].astype(str)
  df[col]=label_encoder.fit_transform(df[col])
```

```python
df.head()
```

```python
df.drop(['Order Date'],axis=1,inplace=True)
```

```python
df.to_excel("Data set.csv.xlsx",index=False)
```

X AND Y JUSTIFICATION


```python
y=df['Sales']
x=df.drop(['Sales', 'Order ID', 'Day Of Week'],axis=1)
```

MULTI LINEAR REGRESSION


```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
model = LinearRegression()
```

```python
model.fit(x_train, y_train)
```

```python
model.score(x_train, y_train)
```

```python
model
```

```python
model.coef_
```

```python
model.intercept_
```

DATA EVALUTION


```python
predict=model.predict(x_test)
predict
```

```python
result=pd.DataFrame({'Actual':y_test,'Predicted':predict})
result
```

```python
mse=mean_squared_error(y_test,predict)
mse
```

```python
mean_absolute_error(y_test,predict)
```

```python
rmse=np.sqrt(mse)
rmse
```

```python
r2_score=r2_score(y_test,predict)
r2_score
```
